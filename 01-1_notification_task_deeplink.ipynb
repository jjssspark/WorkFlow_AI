# 업무 알림 딥링크 구현

## 실행 계획
1. 종 아이콘 알림 목록과 실시간 토스트의 업무 알림을 `/board?taskId=...`에 연결한다.
2. 업무보드 로딩 후 대상 업무의 상세 패널을 자동으로 연다.
3. 활성 필터 때문에 대상 카드가 숨지 않도록 업무 딥링크 진입 시 필터를 초기화한다.
4. 삭제되었거나 없는 업무에는 사용자 안내를 표시한다.
5. 알림 및 업무보드 테스트와 프로덕션 빌드를 검증한다.


## 진행 상태
- [x] 기존 딥링크 처리 확인
- [x] 알림 클릭 연결
- [x] 예외 처리
- [x] 테스트 및 문서화

## 검증 결과
- 알림 토스트·헤더·업무보드 테스트 24개 통과
- 이미 열린 업무보드에서 URL의 taskId만 바뀌는 회귀 시나리오 통과
- Vite 프로덕션 빌드 통과
