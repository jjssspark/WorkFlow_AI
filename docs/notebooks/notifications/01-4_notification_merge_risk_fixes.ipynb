# 프로젝트 알림 병합 충돌 및 위험 보완

## 실행 계획
1. 최신 `dev`를 `feature/kanban`에 병합하고 충돌 파일을 확인한다.
2. 사용자·프로젝트별로 알림 보관 한도를 적용한다.
3. 기존 알림의 `project_id` 역산 범위를 보완하고 역산 불가 데이터의 처리 방침을 명시한다.
4. 진행률 보고서 알림 생성 전에 로그인 사용자의 프로젝트 소속을 검증한다.
5. 관련 백엔드 테스트와 마이그레이션 정적 검증을 실행한다.
