# 프로젝트별 알림 분리 작업

## 목표
현재 사용자 단위로 통합된 알림을 현재 선택한 프로젝트 기준으로 조회·표시한다.

## 실행 계획
1. 알림 엔티티와 DB에 `project_id`를 추가한다.
2. 업무보드 및 프로젝트 기능에서 알림 생성 시 프로젝트 ID를 기록한다.
3. 알림 목록과 미확인 개수 API에 프로젝트 필터를 적용한다.
4. 프론트엔드 조회·SSE 토스트를 현재 프로젝트 기준으로 제한한다.
5. 백엔드·프론트엔드 테스트로 회귀 여부를 확인한다.
