# 완료 승인 요청 알림 바로가기 개선

## 목표
승인 요청 알림의 바로가기 버튼을 누르면 팀장 페이지의 완료승인 대기로 이동하고, 알림이 가리키는 업무의 상세 패널을 자동으로 연다.

## 실행 계획

1. `COMPLETION_REQUESTED` 알림의 `targetType=task`, `targetId` 전달 여부를 확인한다.
2. 알림 유형별 딥링크를 한 함수에서 생성한다.
3. `/leader/completion-approvals?taskId=...` 경로를 추가한다.
4. 승인 대기 목록을 불러온 뒤 해당 업무를 선택해 상세 패널을 연다.
5. 헤더 알림, 실시간 토스트, 승인 대기 화면 테스트를 실행한다.

## 검증 기준

- 승인 요청 알림에 `할 일` 배지와 `바로가기`가 표시된다.
- 헤더와 실시간 토스트가 동일한 딥링크를 사용한다.
- `taskId`가 승인 대기 목록에 있으면 해당 행이 선택되고 상세 패널이 열린다.
- 기존 회의록·평가 알림 바로가기는 회귀하지 않는다.
